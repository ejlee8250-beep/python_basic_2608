## 1. 클래스는 왜 필요한가?

```python
#계산기 프로그램을 함수로 만들기
result = 0

def add(num):
    global result
    result += num
    return result

add(3)
```

```python
add(4)
```

```python
#계산기 2대가 필요한 상황
result1 = 0
result2 = 0

def add1(num):
    global result1
    result1 += num
    return result1


def add2(num):
    global result2
    result2 += num
    return result2

add1(3)
print(add1(4))

add2(3)
print(add2(7))
```

```python
# 클래스로 해결

class Calculator:
    #생성자 함수(클래스 변수 초기화)
    def __init__(self):
        self.result = 0
    
    def add(self, num):
        self.result += num
        return self.result

    def sub(self, num):
        self.result -= num
        return self.result
```

## 4. 생성자(Constructor)

```python
a = FourCal()
a.add()
```

```python
class FourCal:

    #생성자 함수
    def __init__ (self, first, second):
         self.first = first
         self.second = second

    #더하기함수
    def add(self):
        result = self.first + self.second
        return result

    #곱하기 함수
    def mul(self):
        result = self.first * self.second
        return result

    #빼기 함수
    def sub(self):
        result =self.first - self.second
        return result

    #나누기 함수
    def div(self):
        result = self.first / self.second
        return result
    
```

```python
a = FourCal()
```

```python
#FourCal 클래스로 객체가 생성되면 생성자 함수에게 매개변수값이 전달된다
a = FourCal(4,2)
```

```python
a.add()
```

```python
a.mul()
```

```python
a.sub()
```

```python
a.div()
```
