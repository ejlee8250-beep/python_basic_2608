## 2. 자료형의 참과 거짓

자료형에 참과 거짓이 프로그램에서 어떻게 쓰이는지 간단히 알아보자

```python
a = [1,2,3,4]

while a:
    print(a.pop())
```

```python
if [1,2]:
    print('참')
else:
    print('거짓')
```

```python
if []:
    print('참')
else:
    print('거짓')
```
