### random : 난수(규칙이 없는 임의의 수)를 발생

```python
import random

random.random()  # 0<= x <1.0 
```

```python
# randint() : 정수 중에서 난수 발생
random.randint(1,10) # 1<= x <=10 
```

```python
#random_pop함수
def random_pop(data):
    num = random.randint(0,len(data)-1)
    print(data,num)
    return data.pop(num)

data = [1,2,3,4,5]
while data:
    print(random_pop(data))
```

random.choice ():입력받은 리스트에서 무작위로 하나의 요소를 선택하여 리턴

```python
def random_pop(data):
    num = random.choice(data)
    print(data,num)
    data.remove(num)
    return num

data = [1,2,3,4,5]
while data:
    print(random_pop(data))
```

ramdom.sample() : 리스트 요소를 무작위로 섞어서 반환

```python
data = [1,2,3,4,5]
random.sample(data,len(data))
```

pickle : 객체의 형태를 그대로 유지하면서 파일을 저장하고 불러오게 하는 모듈(피클데이터는 바이너리 데이터)

```python
import pickle

with open('test.txt','wb')as f:
    data = {1: 'python',2: 'you need'}
    pickle.dump(data,f)
```

```python
with open('test.txt','rb') as f:
    content = pickle.load(f)
    print(content)
```

json : JSON 데이터 처리할 때 사용

```python
# dict자료형 -> json파일로
data = {'name':'홍길동','birth' : '0525','age' : 30}

with open('myinfo.json','w')as f:
    json.dump(data,f)
```

zipfile : 여러개 파일을 zip형식으로 합치거나 해제

```python
import zipfile

with zipfile.ZipFile('mytest.zip','w') as myzip:
    myzip.write('a.txt')
    myzip.write('b.txt')
    myzip.write('c.txt')
```

```python
# 압축 풀기
with zipfile.ZipFile('mytest.zip') as myzip:
    myzip.extractall()
```
