## 10. format()함수를 사용한 포매팅

```python
"I eat {} apples.".format(3)
```

```python
"I eat {} apples.".format('five')
```

```python
num = 3
"I eat {} apples.".format(num)
```

```python
number = 10
day ='three'

"I ate {0} apples, so I was sick for {1} day.".format(number, day)
```

```python
# 변수명으로 넣기
"I ate {number} apples, so I was sick for {day} day.".format(day= 'three', number=10)
```

```python
# 인덱스와 변수명을 혼합해서 넣기
"I ate {0} apples, so I was sick for {day} day.".format(10, day =3)
```

```python
#왼쪽 정렬
"{:<10}".format('hi')
```

```python
#오른쪽 정렬
"{:>10}".format('hi')
```

```python
#가운데 졍렬
"{:^10}".format('hi')
```

```python
#공백 채우기
"{:=^10}".format('hi')
```

```python
"{:!<10}".format('hi')
```

```python
#소수점 표현
"{:0.4f}".format(3.141592)
```

```python
"{:10.4f}".format(3.141592)
```

```python
"{{and}}".format()
```
