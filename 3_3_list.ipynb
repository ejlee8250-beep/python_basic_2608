## 4.리스트 관련 함수

리스트에 요소 추가 : append(value)

```python
a = [1,2,3]

a.append('four')
a
```

```python
a.append([5,6])
a
```

리스트 정렬 : sort() 

```python
a = [3,7,5,1,4]
a.sort()
a
```

```python
a = ['d','a','c','b']
a.sort()
a
```

```python
a.sort(reverse=True)
a
```

```python
a = ['d','a','c','b']
a.reverse()
a
```
