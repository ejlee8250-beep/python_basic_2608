# 제어문 / 조건문(if문)

```python
# ‘돈이 있으면 택시를 타고 가고, 돈이 없으면 걸어간다.’
moneny = False

if moneny:
    print('택시를 타고 가라')
else:
    print('걸어가라')
```

## 3. 다양한 조건을 판단하는 elif

```python
#if와 else만으로 표현
pocket = ['paper','cellphone']
card = True

if 'moneny' in pocket:
    print('택시를 타고 가라')
else:
    if card:
        print('택시를 타고 가라')
    else:
        print('걸어가라')
    
```

```python
# elif를 사용
if 'moneny' in pocket:
    print('택시를 타고 가라')
elif card:
    print('택시를 타고 가라')
else:
    print('걸어가라')
```

```python
# if문을 한 줄로 작성하기
if 'moneny' in pocket: pass
else: print('카드를 꺼내라')
```
